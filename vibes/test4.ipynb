```python
from itertools import product, combinations, permutations
import time

a = [16, 4, 14, 13, 1,9, 20]
b = ['*', '/', '*', '/']
start = time.time()

def join_numbers(nums):
    """Given a tuple/list of integers, return their string concatenation."""
    return ''.join(map(str, nums))

def is_valid_joined_number(joined, original_nums):
    """
    Return True iff the joined string satisfies:
    1. If it's length > 1, and starts with '0', it's invalid.
    2. If any of the numbers joined is >=10, it must be by itself, i.e. not joined with others.
    """
    # Rule 1: No leading zeros unless it's a single 0 digit
    if len(joined) > 1 and joined[0] == '0':
        return False
    # Rule 2: If any original part is >=10 (2 or more digits), then the join is valid ONLY if exactly one number in the group
    if len(original_nums) > 1 and any(int(x) >= 10 for x in original_nums):
        return False
    return True

def get_all_joined_partitions(operands, max_join=3):
    """
    For a list of operands, generate all possible partitions where up to 'max_join' numbers may be joined.
    Rule: Only single digit numbers can be joined. If any operand >=10, it must stand alone.
    Rule: No joined number can begin with 0 unless it's single '0'
    """
    n = len(operands)
    results = []
    def helper(idx, curr):
        if idx == n:
            results.append(curr[:])
            return
        # Only join up to max_join numbers
        for size in range(1, max_join+1):
            if idx + size <= n:
                curr_group = operands[idx:idx+size]
                joined = join_numbers(curr_group)
                # Rule: If any in curr_group is >=10 and the group size > 1, skip
                if not is_valid_joined_number(joined, curr_group):
                    continue
                # No group should join (join across) a >=10 number
                curr.append(joined)
                helper(idx+size, curr)
                curr.pop()
    helper(0, [])
    return results

def is_valid_expr(expr):
    """
    expr is a string like "12/01*3" or "1/0" etc.
    Rules:
    - No "/0" (cannot divide by zero) where "0" is a standalone zero (single digit)
    """
    return '/0' not in expr

def generate_all_subexprs(operands, operators):
    results = set()
    n = len(operands)
    for subset_size in range(1, n+1):
        for perm in permutations(operands, subset_size):
            joined_lists = get_all_joined_partitions(list(perm), max_join=3)
            for joined in joined_lists:
                # Don't even consider joined stuff with invalid numbers (already checked in joined partition)
                if len(joined) == 1:
                    expr = joined[0]
                    if is_valid_expr(expr):
                        results.add(expr)
                else:
                    ops_needed = len(joined) - 1
                    for ops in set(permutations(operators, ops_needed)):
                        expr_parts = []
                        for i in range(ops_needed):
                            expr_parts.append(str(joined[i]))
                            expr_parts.append(ops[i])
                        expr_parts.append(str(joined[-1]))
                        expr_str = ''.join(expr_parts)
                        if is_valid_expr(expr_str):
                            results.add(expr_str)
    return sorted(results, key=lambda x: (len(x), x))

all_exprs = generate_all_subexprs(a, b)
all_exprs
# print(len(all_exprs))
# print(time.time()-start)


# takes 10^-3 in C with
# takes 10^-2 in python with
# a = [1, 2, 3, 0,0]
# b = ['*', '/', '*', '/']
```

```python
from itertools import combinations, permutations
import time

a = [2, 4, 6, 8, 12]
b = ['*', '/']
start = time.time()

def join_numbers(nums):
    return ''.join(map(str, nums))

def is_valid_joined_number(joined, original_nums):
    if len(joined) > 1 and joined[0] == '0':
        return False
    if len(original_nums) > 1 and any(int(x) >= 10 for x in original_nums):
        return False
    return True

def canonical_outer_pair(s1, s2):
    return (s1, s2) if s1 <= s2 else (s2, s1)

def is_valid_value(value):
    """
    Filter values:
    - Must be <= 1000
    - If has decimal part, decimal must be >= 0.01
    """
    if value > 1000:
        return False
    
    # Check decimal part
    decimal_part = abs(value - int(value))
    
    # If there's a decimal part, it must be >= 0.01
    if decimal_part > 0 and decimal_part < 0.1:
        return False
    
    return True

def generate_term_multisets_from_subset(subset, max_join=3):
    """
    Given an unordered subset of operands (unique ints),
    generate all possible *term multisets* (as sorted tuple of strings),
    where terms are either:
      - a standalone operand >= 10 (must stand alone)
      - a joined number from 1..max_join single-digit operands, with internal order allowed
    This matches what permutations+consecutive-joining can produce, but without factorial blow-up.
    
    Returns: set of tuples of terms (NOT tuples of (terms, _))
    """
    big = [x for x in subset if x >= 10]
    digits = sorted([x for x in subset if 0 <= x <= 9])

    big_terms = [str(x) for x in big]
    out = set()

    # Recursively partition digits into blocks of size 1..max_join (unordered partition)
    def rec(remaining_digits, built_terms):
        if not remaining_digits:
            out.add(tuple(sorted(big_terms + built_terms)))
            return

        first = remaining_digits[0]
        rest = remaining_digits[1:]

        # choose block size
        for size in range(1, max_join + 1):
            if size == 1:
                block = (first,)
                block_rest = rest
                # options for this block
                opts = [str(first)]
            else:
                if len(rest) < size - 1:
                    continue
                # choose the other (size-1) digits to join with 'first' (combinations avoids duplicates)
                for extra in combinations(rest, size - 1):
                    block = (first,) + extra
                    # remove chosen extras from rest
                    extra_set = set(extra)
                    block_rest = [d for d in rest if d not in extra_set]

                    # internal permutations define the joined value
                    opts = []
                    for p in permutations(block):
                        joined = join_numbers(p)
                        if is_valid_joined_number(joined, p):
                            opts.append(joined)
                    if not opts:
                        continue

                    # branch on each valid joined value
                    for joined_str in opts:
                        built_terms.append(joined_str)
                        rec(block_rest, built_terms)
                        built_terms.pop()
                continue  # important: don't fall through for size>1

            # size==1 case
            for joined_str in opts:
                built_terms.append(joined_str)
                rec(block_rest, built_terms)
                built_terms.pop()

    rec(digits, [])
    return out

def generate_all_pairs(operands, operators, max_join=3):
    """
    Return dict: value -> [(term_string, number_sig, operator_sig), ...]
    
    number_sig: binary string showing which number tiles are used
    operator_sig: binary string showing which operator tiles are used
    """
    star_budget = sum(op == '*' for op in operators)
    slash_budget = sum(op == '/' for op in operators)
    
    n_numbers = len(operands)
    n_operators = len(operators)
    
    # Map operands to their indices
    operand_to_indices = {}
    for i, op in enumerate(operands):
        if op not in operand_to_indices:
            operand_to_indices[op] = []
        operand_to_indices[op].append(i)
    
    value_to_terms = {}

    for subset_size in range(1, n_numbers + 1):
        for subset in combinations(operands, subset_size):
            # Get indices of this subset
            subset_indices = []
            temp_operand_to_indices = {k: list(v) for k, v in operand_to_indices.items()}
            for num in subset:
                idx = temp_operand_to_indices[num].pop(0)
                subset_indices.append(idx)

            term_multisets = generate_term_multisets_from_subset(subset, max_join=max_join)

            for terms in term_multisets:  # terms is just a tuple of strings
                L = len(terms)

                # single-term case (no operators)
                if L == 1:
                    term_str = terms[0]
                    try:
                        value = eval(term_str)
                    except Exception:
                        continue

                    if not is_valid_value(value):
                        continue
                    
                    # Create binary signatures
                    num_sig = 0
                    for idx in subset_indices:
                        num_sig |= (1 << idx)
                    op_sig = 0
                    
                    num_sig_str = bin(num_sig)[2:].zfill(n_numbers)[::-1]
                    op_sig_str = bin(op_sig)[2:].zfill(n_operators)[::-1] if n_operators > 0 else ""
                    
                    if value not in value_to_terms:
                        value_to_terms[value] = []
                    value_to_terms[value].append((term_str, num_sig_str, op_sig_str))
                    continue

                ops_needed = L - 1
                if ops_needed > star_budget + slash_budget:
                    continue

                # feasible number of slashes k
                min_slash = max(0, ops_needed - star_budget)
                max_slash = min(ops_needed, slash_budget)

                # choose which terms go to divisor side (k of them), ensuring at least 1 multiplier remains
                # and forbidding '0' as a divisor term
                idxs_ok_for_div = [i for i, t in enumerate(terms) if t != '0']

                for k in range(min_slash, max_slash + 1):
                    if k >= L:
                        continue  # need at least one multiplier term
                    if k > len(idxs_ok_for_div):
                        continue

                    # choose divisor indices
                    for div_idxs in combinations(idxs_ok_for_div, k):
                        div_set = set(div_idxs)
                        divs = [terms[i] for i in div_set]
                        mults = [terms[i] for i in range(L) if i not in div_set]
                        
                        # Calculate operator signature
                        n_stars = ops_needed - k
                        n_slashes = k
                        
                        op_sig = 0
                        # Use first n_stars '*' operators
                        star_count = 0
                        for i, op in enumerate(operators):
                            if op == '*' and star_count < n_stars:
                                op_sig |= (1 << i)
                                star_count += 1
                        # Use first n_slashes '/' operators
                        slash_count = 0
                        for i, op in enumerate(operators):
                            if op == '/' and slash_count < n_slashes:
                                op_sig |= (1 << i)
                                slash_count += 1
                        
                        # Build expression and evaluate
                        expr = '*'.join(mults)
                        if divs:
                            expr += '/' + '/'.join(divs)
                        
                        try:
                            value = eval(expr)
                        except Exception:
                            continue

                        if not is_valid_value(value):
                            continue
                        
                        # Create binary signatures
                        num_sig = 0
                        for idx in subset_indices:
                            num_sig |= (1 << idx)
                        
                        num_sig_str = bin(num_sig)[2:].zfill(n_numbers)[::-1]
                        op_sig_str = bin(op_sig)[2:].zfill(n_operators)[::-1] if n_operators > 0 else ""
                        
                        if value not in value_to_terms:
                            value_to_terms[value] = []
                        value_to_terms[value].append((expr, num_sig_str, op_sig_str))

    return value_to_terms

# Generate with metadata
value_to_terms = generate_all_pairs(a, b, max_join=3)

# Print results
print(f"Generated {len(value_to_terms)} unique values in {time.time() - start:.3f} seconds\n")

# Display first few values
for i, (value, terms) in enumerate(sorted(value_to_terms.items())):
    print(f"Value {value}:")
    for term, num_sig, op_sig in terms[:5]:  # Show first 5 terms per value
        print(f"  {term:20s} | nums: {num_sig} | ops: {op_sig}")
    if len(terms) > 5:
        print(f"  ... ({len(terms) - 5} more terms)")
    print()

print(f"\nTotal execution time: {time.time() - start:.3f} seconds")
```

```python
from fractions import Fraction
from functools import lru_cache

# -----------------------------
# Demo tiles (operands + operators + '=')
# -----------------------------
# Indices: 0..8
tiles = [2, 2, 4, 6, 1, "*", "/", "+", "="]
n = len(tiles)
EQ_BIT = 1 << tiles.index("=")

def is_num(t): return isinstance(t, int)
def is_op(t): return t in {"+", "-", "*", "/"}
def is_eq(t): return t == "="

def mask_bits(mask):
    """Return list of bit indices set in mask."""
    out = []
    i = 0
    while mask:
        if mask & 1:
            out.append(i)
        mask >>= 1
        i += 1
    return out

def describe_mask(mask):
    """Human-readable list of tiles used by mask."""
    bits = mask_bits(mask)
    return [(i, tiles[i]) for i in bits]

def count_nums_ops(mask):
    nums = ops = 0
    for i in mask_bits(mask):
        t = tiles[i]
        if is_eq(t):
            continue
        if is_num(t): nums += 1
        elif is_op(t): ops += 1
    return nums, ops

def is_expr_feasible(mask):
    """
    A mask is a valid expression "shape" if:
      - it doesn't use '='
      - nums >= 1
      - ops == nums - 1    (a binary-expression tree)
    """
    if mask & EQ_BIT:
        return False
    nums, ops = count_nums_ops(mask)
    return nums >= 1 and ops == nums - 1

# -----------------------------
# Subset DP over masks (classic arithmetic DP)
# DP[mask] = dict[value -> one representative expression string]
# -----------------------------
@lru_cache(maxsize=None)
def solve_mask(mask):
    """
    Returns dict: {Fraction -> expr_string} for expressions using exactly tiles in 'mask'.
    """
    if not is_expr_feasible(mask):
        return {}

    # Base case: exactly one number and no operators
    nums, ops = count_nums_ops(mask)
    if nums == 1 and ops == 0:
        for i in mask_bits(mask):
            t = tiles[i]
            if is_num(t):
                v = Fraction(t, 1)
                return {v: str(t)}
        return {}

    results = {}  # value -> expr (keep first seen)

    # Choose which operator tile is the "root" of this expression tree
    for op_i in mask_bits(mask):
        op = tiles[op_i]
        if not is_op(op):
            continue

        rem = mask & ~(1 << op_i)  # remaining tiles to split into left/right subexpressions

        # Enumerate submasks A of rem (non-empty proper subsets)
        # Use symmetry breaking: consider A <= B by numeric mask comparison
        sub = (rem - 1) & rem
        while sub:
            A = sub
            B = rem ^ A
            if A and B and A < B:
                left = solve_mask(A)
                right = solve_mask(B)
                if left and right:
                    for va, ea in left.items():
                        for vb, eb in right.items():
                            # Combine depending on operator
                            if op == "+":
                                v = va + vb
                                if v not in results:
                                    results[v] = f"({ea}+{eb})"
                            elif op == "*":
                                v = va * vb
                                if v not in results:
                                    results[v] = f"({ea}*{eb})"
                            elif op == "-":
                                # Non-commutative: include both orders
                                v1 = va - vb
                                if v1 not in results:
                                    results[v1] = f"({ea}-{eb})"
                                v2 = vb - va
                                if v2 not in results:
                                    results[v2] = f"({eb}-{ea})"
                            elif op == "/":
                                # Non-commutative: include both orders if denom != 0
                                if vb != 0:
                                    v1 = va / vb
                                    if v1 not in results:
                                        results[v1] = f"({ea}/{eb})"
                                if va != 0:
                                    v2 = vb / va
                                    if v2 not in results:
                                        results[v2] = f"({eb}/{ea})"

            sub = (sub - 1) & rem

    return results

# Compute DP for all masks
DP = {}
for mask in range(1 << n):
    if is_expr_feasible(mask):
        DP[mask] = solve_mask(mask)

# -----------------------------
# Printing: show DP results for some masks
# -----------------------------
print("Tiles:")
for i, t in enumerate(tiles):
    print(f"  {i}: {t}")
print()

def show_mask_results(mask, limit=10):
    vals = DP.get(mask, {})
    print(f"mask {mask:09b} ({mask}) uses: {describe_mask(mask)}")
    print(f"  #values: {len(vals)}")
    for k, (v, expr) in enumerate(list(vals.items())[:limit]):
        print(f"    {v}  <-  {expr}")
    print()

# Example masks:
#  - {4} alone
show_mask_results(1 << 2)  # tile '4'
#  - {2,*,2} = indices {0,5,1}? careful: '*' is index 5 in our list
mask_2star2 = (1<<0) | (1<<1) | (1<<5)  # 2 * 2
show_mask_results(mask_2star2)
#  - {4,+,6} is NOT feasible as an expression because '+' needs two operands,
#    so we'd need 4,6 and '+' only => nums=2 ops=1 OK (but '+' is op tile). Let's show it:
mask_4plus6 = (1<<2) | (1<<3) | (1<<7)  # 4,6,'+'
show_mask_results(mask_4plus6)

# -----------------------------
# (4) Fast equation finding: value -> bitset of masks
# Build H[value] = bitset over masks that can produce 'value'
# -----------------------------
H = {}  # Fraction -> python int bitset, bit 'mask' set if mask can make value

for mask, valmap in DP.items():
    for v in valmap.keys():
        H[v] = H.get(v, 0) | (1 << mask)

# Precompute disjointness bitsets D[mask]:
# D[mask] has bit s set iff s is disjoint with mask and is a feasible expression-mask
D = [0] * (1 << n)
feasible_masks_bitset = 0
for m in range(1 << n):
    if is_expr_feasible(m):
        feasible_masks_bitset |= (1 << m)

for m in range(1 << n):
    # disjoint masks are subsets of (~m), but restrict to feasible expression masks
    dis = 0
    for s in range(1 << n):
        if (s & m) == 0 and (feasible_masks_bitset >> s) & 1:
            dis |= (1 << s)
    D[m] = dis

def iter_set_bits(bitset_int):
    """Yield positions of 1 bits in an int bitset."""
    while bitset_int:
        lsb = bitset_int & -bitset_int
        i = (lsb.bit_length() - 1)
        yield i
        bitset_int ^= lsb

# Find equations A = B where:
#  - A and B evaluate to same value v
#  - masks are disjoint
#  - '=' tile is available (in rack). We'll just assume we can place it (it's in tiles).
# For demo, we don't force using all tiles; we find any disjoint pair of expression masks.
print("=== Equations found (A = B) using bitset disjointness ===")

shown = 0
MAX_SHOW = 15

for v, bitset_masks in H.items():
    # Skip if fewer than 2 masks produce v
    if bitset_masks & (bitset_masks - 1) == 0:
        continue

    # For each mask m producing v, find partner masks disjoint with m
    for m in iter_set_bits(bitset_masks):
        partners = bitset_masks & D[m]
        # avoid duplicates by requiring partner > m
        partners &= ~((1 << (m + 1)) - 1)  # clear bits <= m
        for p in iter_set_bits(partners):
            # produce one representative expr for each side
            expr_m = DP[m][v]
            expr_p = DP[p][v]
            print(f"{expr_m} = {expr_p}   (value {v}, masks {m},{p})")
            shown += 1
            if shown >= MAX_SHOW:
                break
        if shown >= MAX_SHOW:
            break
    if shown >= MAX_SHOW:
        break

print()

# Bonus: show how to join a specific value quickly:
target = Fraction(4, 1)
print(f"=== Debug: masks that make value {target} ===")
if target in H:
    ms = list(iter_set_bits(H[target]))
    print("masks:", ms)
    for m in ms[:10]:
        print(f"  mask {m:09b} ({m}): {DP[m][target]}  uses {describe_mask(m)}")
else:
    print("No masks found for that value.")
```

```python
"""
A-Math Equation Generator with Tile Signature Optimization

This demonstrates how to use binary tile signatures to efficiently generate
valid equations while respecting tile constraints.
"""

from collections import defaultdict
from itertools import combinations
import time


class AMathGenerator:
    def __init__(self, tiles):
        """
        Initialize with a tile rack.
        tiles: list of numbers and operators, e.g., [2, 4, 7, 1, '*', '/', '+', '=']
        """
        self.tiles = tiles
        self.rack_size = len(tiles)
        self.tile_index = {}  # signature -> {value: [terms]}
        
    def make_signature(self, indices):
        """Convert list of tile indices to binary signature."""
        sig = 0
        for idx in indices:
            sig |= (1 << idx)
        return sig
    
    def sig_to_string(self, sig):
        """Convert signature to binary string (reversed for readability)."""
        return bin(sig)[2:].zfill(self.rack_size)[::-1]
    
    def sig_to_tiles(self, sig):
        """Extract which tiles are used in a signature."""
        used = []
        for i in range(self.rack_size):
            if sig & (1 << i):
                used.append(self.tiles[i])
        return used
    
    def popcount(self, sig):
        """Count number of bits set in signature."""
        return bin(sig).count('1')
    
    def add_to_index(self, sig, value, term):
        """Add a term to the tile index."""
        if sig not in self.tile_index:
            self.tile_index[sig] = defaultdict(list)
        if term not in self.tile_index[sig][value]:
            self.tile_index[sig][value].append(term)
    
    def build_index(self):
        """Build the complete tile signature index."""
        print("=" * 60)
        print("STEP 1: Building Tile Signature Index")
        print("=" * 60)
        print(f"Tiles: {self.tiles}\n")
        
        # Step 1a: Single tiles
        print("1a. Single tile terms:")
        for i, tile in enumerate(self.tiles):
            if isinstance(tile, (int, float)):
                sig = 1 << i
                value = tile
                term = str(tile)
                self.add_to_index(sig, value, term)
                print(f"  Tile[{i}]={tile} → sig={self.sig_to_string(sig)} "
                      f"value={value} term='{term}'")
        
        # Step 1b: Joined numbers (2-3 digits)
        print("\n1b. Joined number terms:")
        digit_indices = [(i, t) for i, t in enumerate(self.tiles) 
                        if isinstance(t, int) and 0 <= t <= 9]
        
        # Two-digit numbers
        for i in range(len(digit_indices)):
            for j in range(len(digit_indices)):
                if i == j:
                    continue
                idx1, d1 = digit_indices[i]
                idx2, d2 = digit_indices[j]
                
                # Skip leading zeros
                if d1 == 0:
                    continue
                    
                joined = int(f"{d1}{d2}")
                sig = self.make_signature([idx1, idx2])
                term = f"{d1}{d2}"
                self.add_to_index(sig, joined, term)
                print(f"  Tiles[{idx1},{idx2}]={d1},{d2} → sig={self.sig_to_string(sig)} "
                      f"value={joined} term='{term}'")
        
        # Three-digit numbers
        for i in range(len(digit_indices)):
            for j in range(len(digit_indices)):
                for k in range(len(digit_indices)):
                    if i == j or i == k or j == k:
                        continue
                    idx1, d1 = digit_indices[i]
                    idx2, d2 = digit_indices[j]
                    idx3, d3 = digit_indices[k]
                    
                    if d1 == 0:
                        continue
                        
                    joined = int(f"{d1}{d2}{d3}")
                    sig = self.make_signature([idx1, idx2, idx3])
                    term = f"{d1}{d2}{d3}"
                    self.add_to_index(sig, joined, term)
                    print(f"  Tiles[{idx1},{idx2},{idx3}]={d1},{d2},{d3} → "
                          f"sig={self.sig_to_string(sig)} value={joined} term='{term}'")
        
        # Step 1c: Multiplication and division
        print("\n1c. Multiplication and Division terms:")
        
        # Find operator indices
        op_indices = {}
        for i, tile in enumerate(self.tiles):
            if tile in ['*', '/', '+', '-', '=']:
                if tile not in op_indices:
                    op_indices[tile] = []
                op_indices[tile].append(i)
        
        print(f"  Available operators: {op_indices}")
        
        # Get all existing number signatures
        number_sigs = list(self.tile_index.keys())
        
        # Try combining numbers with operators
        for op in ['*', '/']:
            if op not in op_indices:
                continue
            
            for op_idx in op_indices[op]:
                op_sig = 1 << op_idx
                
                for sig1 in number_sigs[:]:
                    for sig2 in number_sigs[:]:
                        # Check for overlaps
                        if sig1 & sig2:
                            continue
                        if (sig1 & op_sig) or (sig2 & op_sig):
                            continue
                        
                        new_sig = sig1 | sig2 | op_sig
                        
                        for val1_str, terms1 in self.tile_index[sig1].items():
                            for val2_str, terms2 in self.tile_index[sig2].items():
                                val1 = float(val1_str)
                                val2 = float(val2_str)
                                
                                # Apply operation
                                if op == '*':
                                    new_value = val1 * val2
                                elif op == '/' and val2 != 0:
                                    new_value = val1 / val2
                                else:
                                    continue
                                
                                for term1 in terms1[:2]:  # Limit combinations
                                    for term2 in terms2[:2]:
                                        new_term = f"{term1}{op}{term2}"
                                        self.add_to_index(new_sig, new_value, new_term)
                                        
                                        val_display = f"{new_value:.2f}" if new_value != int(new_value) else str(int(new_value))
                                        print(f"  {term1} {op} {term2} → sig={self.sig_to_string(new_sig)} "
                                              f"value={val_display} term='{new_term}'")
    
    def print_index_summary(self):
        """Print summary of the tile index."""
        print("\n" + "=" * 60)
        print("STEP 2: Tile Signature Index Summary")
        print("=" * 60)
        
        sorted_sigs = sorted(self.tile_index.keys())
        
        print(f"\nTotal unique signatures: {len(sorted_sigs)}\n")
        
        for i, sig in enumerate(sorted_sigs):
            binary = self.sig_to_string(sig)
            used_tiles = self.sig_to_tiles(sig)
            
            print(f"Signature {i+1}: {binary} [tiles: {used_tiles}]")
            
            values = sorted(self.tile_index[sig].keys(), 
                          key=lambda x: float(x))
            
            for val in values:
                terms = self.tile_index[sig][val]
                val_display = f"{float(val):.2f}" if float(val) != int(float(val)) else str(int(float(val)))
                terms_display = ', '.join(terms)
                print(f"  Value {val_display}: {terms_display}")
            
            print()
    
    def find_equations_v1_equals_v2(self):
        """Find equations of type V1 = V2."""
        print("=" * 60)
        print("STEP 3a: Finding Equations V1 = V2")
        print("=" * 60)
        print()
        
        # Find = operator
        equals_indices = [i for i, tile in enumerate(self.tiles) if tile == '=']
        
        if not equals_indices:
            print("No '=' operator available\n")
            return []
        
        equations = []
        sorted_sigs = sorted(self.tile_index.keys())
        
        for equals_idx in equals_indices:
            equals_sig = 1 << equals_idx
            
            for i, sig1 in enumerate(sorted_sigs):
                for sig2 in sorted_sigs[i+1:]:
                    # Must be disjoint (no overlapping tiles)
                    if sig1 & sig2:
                        continue
                    # Must not use the = operator
                    if (sig1 & equals_sig) or (sig2 & equals_sig):
                        continue
                    
                    # Find matching values
                    for val in self.tile_index[sig1].keys():
                        if val in self.tile_index[sig2]:
                            for term1 in self.tile_index[sig1][val][:2]:
                                for term2 in self.tile_index[sig2][val][:2]:
                                    equations.append({
                                        'type': 'V1=V2',
                                        'equation': f"{term1} = {term2}",
                                        'sig1': sig1,
                                        'sig2': sig2,
                                        'equals_sig': equals_sig,
                                        'value': val
                                    })
                                    
                                    print(f"Equation: {term1} = {term2}")
                                    print(f"  LHS signature: {self.sig_to_string(sig1)}")
                                    print(f"  RHS signature: {self.sig_to_string(sig2)}")
                                    print(f"  =   signature: {self.sig_to_string(equals_sig)}")
                                    print(f"  All disjoint: {(sig1 & sig2 & equals_sig) == 0}")
                                    print(f"  Value: {val}")
                                    print()
        
        print(f"\nTotal V1=V2 equations: {len(equations)}\n")
        return equations
    
    def find_equations_v1_plus_v2_equals_v3(self):
        """Find equations of type V1 + V2 = V3."""
        print("=" * 60)
        print("STEP 3b: Finding Equations V1 + V2 = V3")
        print("=" * 60)
        print()
        
        # Find + and = operators
        plus_indices = [i for i, tile in enumerate(self.tiles) if tile == '+']
        equals_indices = [i for i, tile in enumerate(self.tiles) if tile == '=']
        
        if not plus_indices:
            print("No '+' operator available\n")
            return []
        
        if not equals_indices:
            print("No '=' operator available\n")
            return []
        
        equations = []
        sorted_sigs = sorted(self.tile_index.keys())
        
        for plus_idx in plus_indices:
            for equals_idx in equals_indices:
                if plus_idx == equals_idx:
                    continue
                    
                plus_sig = 1 << plus_idx
                equals_sig = 1 << equals_idx
                
                for sig1 in sorted_sigs:
                    for sig2 in sorted_sigs:
                        # Check disjoint and don't use + or = operators yet
                        if sig1 & sig2:
                            continue
                        if (sig1 & plus_sig) or (sig2 & plus_sig):
                            continue
                        if (sig1 & equals_sig) or (sig2 & equals_sig):
                            continue
                        
                        combined_sig = sig1 | sig2 | plus_sig
                        
                        # Look for sig3 that is disjoint from combined and doesn't use =
                        for sig3 in sorted_sigs:
                            if sig3 & combined_sig:
                                continue
                            if sig3 & equals_sig:
                                continue
                            
                            # Try to find matching values
                            for val1_str in list(self.tile_index[sig1].keys())[:5]:
                                for val2_str in list(self.tile_index[sig2].keys())[:5]:
                                    val1 = float(val1_str)
                                    val2 = float(val2_str)
                                    target_val = val1 + val2
                                    
                                    # Check if target value exists in sig3
                                    target_str = str(target_val) if target_val != int(target_val) else str(int(target_val))
                                    
                                    if target_str in self.tile_index[sig3]:
                                        for term1 in self.tile_index[sig1][val1_str][:1]:
                                            for term2 in self.tile_index[sig2][val2_str][:1]:
                                                for term3 in self.tile_index[sig3][target_str][:1]:
                                                    equations.append({
                                                        'type': 'V1+V2=V3',
                                                        'equation': f"{term1} + {term2} = {term3}",
                                                        'sig1': sig1,
                                                        'sig2': sig2,
                                                        'sig3': sig3,
                                                        'plus_sig': plus_sig,
                                                        'equals_sig': equals_sig
                                                    })
                                                    
                                                    print(f"Equation: {term1} + {term2} = {term3}")
                                                    print(f"  V1 signature: {self.sig_to_string(sig1)}")
                                                    print(f"  V2 signature: {self.sig_to_string(sig2)}")
                                                    print(f"  +  signature: {self.sig_to_string(plus_sig)}")
                                                    print(f"  V3 signature: {self.sig_to_string(sig3)}")
                                                    print(f"  =  signature: {self.sig_to_string(equals_sig)}")
                                                    print(f"  All disjoint: {not (sig1 & sig2 & sig3 & plus_sig & equals_sig)}")
                                                    print()
        
        print(f"\nTotal V1+V2=V3 equations: {len(equations)}\n")
        return equations
    
    def print_summary(self, all_equations):
        """Print final summary and complexity analysis."""
        print("=" * 60)
        print("FINAL SUMMARY")
        print("=" * 60)
        print()
        
        v1_v2 = [e for e in all_equations if e['type'] == 'V1=V2']
        v1_v2_v3 = [e for e in all_equations if e['type'] == 'V1+V2=V3']
        
        print(f"Total unique signatures: {len(self.tile_index)}")
        print(f"Total equations found: {len(all_equations)}")
        print(f"  - V1 = V2: {len(v1_v2)}")
        print(f"  - V1 + V2 = V3: {len(v1_v2_v3)}")
        
        print("\n" + "=" * 60)
        print("KEY INSIGHTS")
        print("=" * 60)
        print()
        print("1. Tile signatures prevent impossible combinations")
        print("   - Binary AND checks overlap in O(1) time")
        print("   - No need to check if 2*2 and 2*2/1 are compatible")
        print()
        print("2. Index by signature, not by value alone")
        print("   - Avoids n^k term comparisons")
        print("   - Only valid partitions are checked")
        print()
        
        num_sigs = len(self.tile_index)
        original_complexity = num_sigs ** 3
        optimized_complexity = num_sigs ** 2
        
        print("3. Complexity comparison:")
        print(f"   - Original: O(n^k) = O({num_sigs}^3) ≈ {original_complexity:,} checks")
        print(f"   - Optimized: O(s^2) where s=signatures ≈ {optimized_complexity:,} checks")
        if original_complexity > 0 and optimized_complexity > 0:
            speedup = original_complexity / optimized_complexity
            print(f"   - Speedup: ~{speedup:.0f}x")


def main():
    """Main execution."""
    print("\n" + "=" * 60)
    print("A-MATH EQUATION GENERATOR WITH TILE SIGNATURES")
    print("=" * 60)
    print()
    
    # Hardcoded tile rack - note: now includes '='
    tiles = [1, 2, 4, 6, 8, '*', '/', '+', '*', '=']
    
    print(f"Initializing with tiles: {tiles}")
    print()
    
    start_time = time.time()
    
    # Create generator
    gen = AMathGenerator(tiles)
    
    # Build the index
    gen.build_index()
    
    # Print index summary
    gen.print_index_summary()
    
    # Find equations
    equations_v1_v2 = gen.find_equations_v1_equals_v2()
    equations_v1_v2_v3 = gen.find_equations_v1_plus_v2_equals_v3()
    
    all_equations = equations_v1_v2 + equations_v1_v2_v3
    
    # Print summary
    gen.print_summary(all_equations)
    
    elapsed = time.time() - start_time
    print(f"\nTotal execution time: {elapsed:.3f} seconds")
    print()


if __name__ == "__main__":
    main()
```
